```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries loaded successfully!")
```

```python
df = pd.read_csv("../dataset/sales_data.csv")

df.head()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])

print(df["Order Date"].dtype)
```

```python
df["Year"] = df["Order Date"].dt.year

df[["Order Date", "Year"]].head()
```

```python
df["Month"] = df["Order Date"].dt.month

df[["Order Date", "Year", "Month"]].head()
```

```python
yearly_sales = df.groupby("Year")["Sales"].sum()

yearly_sales
```

```python
monthly_sales = df.groupby("Month")["Sales"].sum()

monthly_sales
```

```python
monthly_sales = (
    df.set_index("Order Date")
      .resample("ME")["Sales"]
      .sum()
)

monthly_sales.head(12)
```

```python
plt.figure(figsize=(14, 6))

plt.plot(monthly_sales.index, monthly_sales.values)

plt.title("Monthly Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(14, 6))

plt.plot(monthly_sales.index, monthly_sales.values)

plt.title("Monthly Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.grid(True)

plt.savefig("../charts/monthly_sales_trend.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

category_sales
```

```python
plt.figure(figsize=(10, 6))

category_sales.plot(kind="bar")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.grid(axis="y")

plt.savefig("../charts/sales_by_category.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)

region_sales
```

```python
plt.figure(figsize=(10, 6))

region_sales.plot(kind="bar")

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.grid(axis="y")

plt.savefig("../charts/sales_by_region.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
segment_sales = df.groupby("Segment")["Sales"].sum().sort_values(ascending=False)

segment_sales
```

```python
plt.figure(figsize=(10, 6))

segment_sales.plot(kind="bar")

plt.title("Sales by Customer Segment")
plt.xlabel("Segment")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.grid(axis="y")

plt.savefig("../charts/sales_by_segment.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
top_products = (
    df.groupby("Product Name")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_products
```

```python
forecast_df = monthly_sales.reset_index()

forecast_df.columns = ["Date", "Sales"]

forecast_df.head()
```

```python
print("Rows:", forecast_df.shape[0])
print("Columns:", forecast_df.shape[1])
```

```python
forecast_df.info()
```

```python
forecast_df["Time_Index"] = np.arange(len(forecast_df))

forecast_df.head()
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    forecast_df["Date"],
    forecast_df["Sales"],
    marker="o"
)

plt.title("Historical Monthly Sales")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.grid(True)

plt.show()
```

```python
forecast_df.to_csv(
    "../dataset/monthly_sales_forecasting.csv",
    index=False
)

print("Forecasting dataset saved successfully!")
```

```python
forecast_df.head()
```

```python
X = forecast_df[["Time_Index"]]
y = forecast_df["Sales"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
train_size = int(len(forecast_df) * 0.8)

X_train = X.iloc[:train_size]
X_test = X.iloc[train_size:]

y_train = y.iloc[:train_size]
y_test = y.iloc[train_size:]

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
print("Training period:")
print(forecast_df.iloc[0]["Date"], "to", forecast_df.iloc[train_size - 1]["Date"])

print()

print("Testing period:")
print(forecast_df.iloc[train_size]["Date"], "to", forecast_df.iloc[-1]["Date"])
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    forecast_df.iloc[:train_size]["Date"],
    y_train,
    label="Training Data"
)

plt.plot(
    forecast_df.iloc[train_size:]["Date"],
    y_test,
    label="Testing Data"
)

plt.title("Training and Testing Data Split")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)

plt.savefig(
    "../charts/train_test_split.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
from sklearn.linear_model import LinearRegression
```

```python
from sklearn.linear_model import LinearRegression

print("Linear Regression imported successfully!")
```

```python
model = LinearRegression()

print("Linear Regression model created!")
```

```python
model.fit(X_train, y_train)

print("Model trained successfully!")
```

```python
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
y_pred = model.predict(X_test)

print("Predictions generated successfully!")
```

```python
comparison = pd.DataFrame({
    "Date": forecast_df.iloc[train_size:]["Date"],
    "Actual_Sales": y_test.values,
    "Predicted_Sales": y_pred
})

comparison
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    comparison["Date"],
    comparison["Actual_Sales"],
    marker="o",
    label="Actual Sales"
)

plt.plot(
    comparison["Date"],
    comparison["Predicted_Sales"],
    marker="o",
    label="Predicted Sales"
)

plt.title("Actual vs Predicted Sales")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)

plt.savefig(
    "../charts/actual_vs_predicted_sales.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

print("Evaluation metrics imported successfully!")
```

```python
mae = mean_absolute_error(y_test, y_pred)

print("Mean Absolute Error (MAE):", round(mae, 2))
```

```python
mse = mean_squared_error(y_test, y_pred)

print("Mean Squared Error (MSE):", round(mse, 2))
```

```python
rmse = np.sqrt(mse)

print("Root Mean Squared Error (RMSE):", round(rmse, 2))
```

```python
r2 = r2_score(y_test, y_pred)

print("R² Score:", round(r2, 4))
```

```python
metrics = pd.DataFrame({
    "Metric": ["MAE", "MSE", "RMSE", "R² Score"],
    "Value": [mae, mse, rmse, r2]
})

metrics
```

```python
metrics.to_csv(
    "../reports/model_evaluation.csv",
    index=False
)

print("Model evaluation saved successfully!")
```
